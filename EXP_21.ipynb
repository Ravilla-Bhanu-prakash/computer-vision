```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()
image_path = list(uploaded.keys())[0]

# Read image
image = cv2.imread(image_path)

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Laplacian mask with diagonal extensions
mask = np.array([
    [1,  1,  1],
    [1, -8,  1],
    [1,  1,  1]
], dtype=np.float32)

# Apply Laplacian mask
laplacian = cv2.filter2D(gray, cv2.CV_32F, mask)

# Sharpen image
sharpened = cv2.subtract(gray.astype(np.float32), laplacian)

# Convert to 8-bit
sharpened = cv2.convertScaleAbs(sharpened)

# Display images
print("Original Image:")
cv2_imshow(image)

print("Sharpened Image:")
cv2_imshow(sharpened)

# Save output
cv2.imwrite("laplacian_diagonal_sharpened.jpg", sharpened)

print("Output saved as laplacian_diagonal_sharpened.jpg")
```
