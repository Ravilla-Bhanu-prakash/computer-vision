```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()
image_path = list(uploaded.keys())[0]

# Read image
image = cv2.imread(image_path)

# Watermark text
text = "WATERMARK"

# Position
position = (20, 50)

# Font
font = cv2.FONT_HERSHEY_SIMPLEX

# Add watermark
cv2.putText(image, text, position, font, 1.5, (255, 255, 255), 3)

# Display image
print("Watermarked Image:")
cv2_imshow(image)

# Save image
cv2.imwrite("watermarked_image.jpg", image)

print("Watermarked image saved as watermarked_image.jpg")

# Download
files.download("watermarked_image.jpg")
```
