```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload an image
uploaded = files.upload()

# Get the uploaded file name
image_path = list(uploaded.keys())[0]

# Read the image
image = cv2.imread(image_path)

# Scale to bigger size (2x)
bigger = cv2.resize(image, None, fx=2, fy=2, interpolation=cv2.INTER_CUBIC)

# Scale to smaller size (0.5x)
smaller = cv2.resize(image, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_AREA)

# Display images
print("Original Image:")
cv2_imshow(image)

print("Bigger Image:")
cv2_imshow(bigger)

print("Smaller Image:")
cv2_imshow(smaller)

# Save the output images
cv2.imwrite("bigger_image.jpg", bigger)
cv2.imwrite("smaller_image.jpg", smaller)

print("Images saved as 'bigger_image.jpg' and 'smaller_image.jpg'")
```
