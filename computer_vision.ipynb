```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload an image
uploaded = files.upload()

# Get the uploaded file name
image_path = next(iter(uploaded))

# Read the image
image = cv2.imread(image_path)

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Display the original image
print("Original Image:")
cv2_imshow(image)

# Display the grayscale image
print("Grayscale Image:")
cv2_imshow(gray)

# Save the grayscale image
cv2.imwrite("grayscale_image.jpg", gray)

print("Grayscale image saved as 'grayscale_image.jpg'")
```

```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload an image
uploaded = files.upload()

# Get the uploaded file name
image_path = next(iter(uploaded))

# Read the image
image = cv2.imread(image_path)

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Apply Gaussian Blur to reduce noise (optional but recommended)
blur = cv2.GaussianBlur(gray, (5, 5), 0)

# Apply Canny Edge Detection
edges = cv2.Canny(blur, 100, 200)

# Display the original image
print("Original Image:")
cv2_imshow(image)

# Display the edge-detected image
print("Canny Edge Detection:")
cv2_imshow(edges)

# Save the output image
cv2.imwrite("canny_edges.jpg", edges)

print("Edge-detected image saved as 'canny_edges.jpg'")
```

```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload an image
uploaded = files.upload()

# Get the uploaded file name
image_path = next(iter(uploaded))

# Read the image
image = cv2.imread(image_path)

# Create a 5x5 kernel
kernel = np.ones((5, 5), np.uint8)

# Apply dilation
dilated = cv2.dilate(image, kernel, iterations=1)

# Display the original image
print("Original Image:")
cv2_imshow(image)

# Display the dilated image
print("Dilated Image:")
cv2_imshow(dilated)

# Save the output image
cv2.imwrite("dilated_image.jpg", dilated)

print("Dilated image saved as 'dilated_image.jpg'")
```

```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload an image
uploaded = files.upload()

# Get the uploaded file name
image_path = next(iter(uploaded))

# Read the image
image = cv2.imread(image_path)

# Create a 5x5 kernel
kernel = np.ones((5, 5), np.uint8)

# Apply erosion
eroded = cv2.erode(image, kernel, iterations=1)

# Display the original image
print("Original Image:")
cv2_imshow(image)

# Display the eroded image
print("Eroded Image:")
cv2_imshow(eroded)

# Save the output image
cv2.imwrite("eroded_image.jpg", eroded)

print("Eroded image saved as 'eroded_image.jpg'")
```
